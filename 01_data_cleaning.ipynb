```python
import pandas as pd
import numpy as np
```

```python
# 1.  Load the dataset
df = pd.read_csv(r"C:\Users\annus\Employee_HR.csv")
```

```python
# --- Step1: Standardize Column Names ---
df.columns = df.columns.str.strip().str.lower()
df.rename(columns={
    'empid': 'emp_id',
    'average_montly_hours': 'average_monthly_hours',
    'salary_inr': 'salary_inr',
    'churn': 'left'
}, inplace=True)
```

```python
# --- Step 2: Handle Invalid / Corrupted Data in Salary ---
df['salary_inr'] = pd.to_numeric(df['salary_inr'], errors='coerce')

median_salary = df.groupby("department")['salary_inr'].transform('median')
df['salary_inr'] = df['salary_inr'].fillna(median_salary)
```

```python
# --- Step 3: Scale/Normalize Scores ---
if df['satisfaction'].max() > 1.0:
    df['satisfaction'] = df['satisfaction'] / 10.0
if df['evaluation'].max() > 1.0:
    df['evaluation'] = df['evaluation'] / 10.0
```

```python
# --- Step 4: Remove Duplicates &  Handle Missing Values ---
df.drop_duplicates(subset=['emp_id'], keep='first', inplace=True)

df.dropna(inplace=True)
```

```python
# --- Step 5: Fix Data Types ---
# Convert Binary flagsto integer/boolean categorical types
categorical_cols = ['work_accident', 'promotion', 'left']
for cols in categorical_cols:
    if cols in df.columns:
        df[cols] = df[cols].astype(int)

# Convert string columns to category type 
df['department'] = df['department'].astype('category')
df['emp_id'] = df['emp_id'].astype(str)
```

```python
# --- Step 6: Save Cleaned Dataset ---
df.to_csv('Cleaned_Employee_HR.csv', index=False)
print("Data Cleaning Completed!")
print("\nDataset Info:")
print(df.info())
print("\nFirst 5 Rows:")
print(df.head())
```
